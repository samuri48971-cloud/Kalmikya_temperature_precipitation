## Построение и анализ графиков температуры и осадков с метеостанции Элиста за 2023 - 2024 гг.

В данной работе будут созданы и рассмотрены два графика: график температур и график осадков за 2023 - 2024 годы. Калмыкия славится своим резко-континентальным климатом, сухими степями и выраженными сезонными контрастами (жарким летом и холодной, малоснежной зимой со значительными колебаниями температур). Для детальной оценки этих особенностей были обработаны суточные метеоданные по городу Элисте.

Для исследования я использовал два инструмента:
1) **numpy** - используется для работы с массивами и числами
2) **matplotlib** - используется для построения графиков

```python
import matplotlib.pyplot as plt
import numpy as np

file = np.loadtxt("weatherElistanew.txt", skiprows = 1, delimiter=",") # загрузка данных
years = file[:, 0] # извлечение столбца с годами
months = file[:, 1] # извлечение столбца с месяцами
days = file[:, 2] # извлечение столбца с днями
temp_day = file[:, 3] # извелечение столбца с дневной температурой
temp_night = file[:, 4] # извлечение столбца с ночной температурой
precipitation = file[:, 5] # извелечение столбца с данными об осадках

time = years + (months - 1) / 12 + (days - 1) / 365 # временная шкала

index_minimum = np.argmin(temp_night) # поиск индекса минимальной ночной температуры
min_temp = temp_night[index_minimum] # сохранение значения минимальной температуры
index_maximum = np.argmax(temp_day) # поиск значения максимальной дневной температуры
max_temp = temp_day[index_maximum] # сохранение значения максимальной температуры

plt.figure(figsize=(14,5)) # создание холста графика для графика 
plt.plot(time, temp_day, label = 'Дневная температура', color = 'black') # построение линии дневной температуры 
plt.plot(time, temp_night, label = 'Ночная температура', color = 'blue') # построение линии ночной температуры
plt.ylabel('Температура °C', fontsize = 18) # добавление подписи к оси y
plt.xlabel('Год, месяц', fontsize = 18) # добавлене подписик к оси x 
plt.grid(True) # сетка значений на фоне
plt.legend(fontsize = 12) # добавление легенды
plt.title('Динамика температур', fontsize = 16) # установка названия графика

plt.show() # вывод графика


plt.figure(figsize=(14,5))
plt.plot(time, precipitation, label = 'Осадки', color = 'violet')
plt.ylabel('Осадки, мм', fontsize = 18)
plt.xlabel('Год, месяц', fontsize = 18)
plt.grid(True)
plt.legend(fontsize = 12)
plt.title('Количество осадков', fontsize = 16)

plt.show()

print(f'Минимальная ночная температура: {min_temp} °C')
print(f'Максимальная дневная температура: {max_temp} °C')
```

## Вывод
На основе анализа графика температур можно сделать вывод, что годовая амплитуда температур в городе Элисте является значительной. Летом температуры достигуют максимальных значений **38 °C** (в конце мая и начале июня), а зимой опускаются до **-22 °C** (в середине января). Из-за малого количества осадков - их максимальное значение за исследуемый период не превысило 16 мм, а дни с дождями бывали редко, что характерно
